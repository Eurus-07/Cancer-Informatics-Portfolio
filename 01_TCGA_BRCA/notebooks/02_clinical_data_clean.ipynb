```python
primary_flag = (
    clinical_unique_records["is_primary_diagnosis"]
    .astype(str)
    .str.lower()
    .eq("true")
)

patients_with_primary = (
    clinical_unique_records.loc[primary_flag, "patient_id"]
    .nunique()
)

total_patients = clinical_unique_records["patient_id"].nunique()

print("Total patients:", total_patients)
print("Patients with primary diagnosis flag:", patients_with_primary)
print(
    "Patients without primary diagnosis flag:",
    total_patients - patients_with_primary
)
```

```python
patients_all = set(
    clinical_unique_records["patient_id"].dropna().unique()
)

patients_with_primary_set = set(
    clinical_unique_records.loc[
        primary_flag,
        "patient_id"
    ].dropna().unique()
)

patients_without_primary = sorted(
    patients_all - patients_with_primary_set
)

patients_without_primary
```

```python
clinical_unique_records[
    clinical_unique_records["patient_id"].isin(
        patients_without_primary
    )
][[
    "patient_id",
    "is_primary_diagnosis",
    "primary_diagnosis",
    "age_at_diagnosis_days",
    "stage",
    "pathologic_T",
    "pathologic_N",
    "pathologic_M",
    "vital_status",
    "days_to_death",
    "days_to_last_follow_up"
]].sort_values("patient_id")
```

```python
primary_records = clinical_unique_records[
    primary_flag
].copy()

print("Rows:", len(primary_records))
print("Unique patients:", primary_records["patient_id"].nunique())
```

```python
fallback_records = clinical_unique_records[
    clinical_unique_records["patient_id"].isin(
        patients_without_primary
    )
].copy()

print("Fallback rows:", len(fallback_records))
print("Fallback patients:", fallback_records["patient_id"].nunique())
```

```python
clinical_patient = pd.concat(
    [primary_records, fallback_records],
    ignore_index=True
)

print("Rows:", len(clinical_patient))
print(
    "Unique patients:",
    clinical_patient["patient_id"].nunique()
)
print(
    "Duplicated patient IDs:",
    clinical_patient["patient_id"].duplicated().sum()
)
```

```python
followup_file = clinical_dir / "follow_up.tsv"

followup_df = pd.read_csv(
    followup_file,
    sep="\t",
    low_memory=False
)

print("Shape:", followup_df.shape)
```

```python
followup_df.columns.tolist()
```

```python
followup_relevant_columns = [
    col for col in followup_df.columns
    if any(
        keyword in col.lower()
        for keyword in [
            "submitter",
            "case_id",
            "follow",
            "death",
            "vital",
            "progression"
        ]
    )
]

followup_relevant_columns
```

```python
followup_selected = followup_df[
    [
        "cases.submitter_id",
        "follow_ups.days_to_follow_up"
    ]
].copy()

followup_selected = followup_selected.rename(
    columns={
        "cases.submitter_id": "patient_id",
        "follow_ups.days_to_follow_up": "days_to_follow_up"
    }
)

followup_selected.head(10)
```

```python
followup_selected["days_to_follow_up"] = pd.to_numeric(
    followup_selected["days_to_follow_up"],
    errors="coerce"
)
```

```python
print("Total follow-up rows:", len(followup_selected))
print(
    "Unique patients:",
    followup_selected["patient_id"].nunique()
)
print(
    "Non-missing follow-up times:",
    followup_selected["days_to_follow_up"].notna().sum()
)
```

```python
followup_selected["days_to_follow_up"].describe()
```

```python
latest_followup = (
    followup_selected
    .dropna(subset=["patient_id", "days_to_follow_up"])
    .groupby("patient_id", as_index=False)["days_to_follow_up"]
    .max()
)

latest_followup = latest_followup.rename(
    columns={
        "days_to_follow_up": "latest_follow_up_days"
    }
)

latest_followup.head()
```

```python
print("Patients with usable follow-up:",
      latest_followup["patient_id"].nunique())

latest_followup["latest_follow_up_days"].describe()
```

```python
latest_followup[
    latest_followup["latest_follow_up_days"] < 0
]
```

```python
(latest_followup["latest_follow_up_days"] < 0).sum()
```

```python
latest_followup.loc[
    latest_followup["latest_follow_up_days"] < 0,
    "latest_follow_up_days"
] = np.nan

print(
    "Negative follow-up values:",
    (latest_followup["latest_follow_up_days"] < 0).sum()
)

print(
    "Patients with usable follow-up:",
    latest_followup["latest_follow_up_days"].notna().sum()
)
```

```python
clinical_patient = clinical_patient.merge(
    latest_followup,
    on="patient_id",
    how="left"
)

print("Rows:", len(clinical_patient))
print("Unique patients:", clinical_patient["patient_id"].nunique())
print(
    "Duplicated patient IDs:",
    clinical_patient["patient_id"].duplicated().sum()
)

print(
    "Missing latest follow-up:",
    clinical_patient["latest_follow_up_days"].isna().sum()
)
```

```python
clinical_patient["vital_status"].value_counts(dropna=False)
```

```python
clinical_patient.groupby("vital_status")[
    ["days_to_death", "latest_follow_up_days"]
].count()
```

```python
clinical_patient["days_to_death"] = pd.to_numeric(
    clinical_patient["days_to_death"],
    errors="coerce"
)
```

```python
clinical_patient["overall_survival_event"] = np.select(
    [
        clinical_patient["vital_status"].eq("Dead"),
        clinical_patient["vital_status"].eq("Alive")
    ],
    [
        1,
        0
    ],
    default=np.nan
)
```

```python
clinical_patient["overall_survival_event"].value_counts(
    dropna=False
)
```

```python
clinical_patient["overall_survival_days"] = np.where(
    clinical_patient["vital_status"].eq("Dead"),
    clinical_patient["days_to_death"],
    np.where(
        clinical_patient["vital_status"].eq("Alive"),
        clinical_patient["latest_follow_up_days"],
        np.nan
    )
)
```

```python
clinical_patient[
    [
        "patient_id",
        "vital_status",
        "days_to_death",
        "latest_follow_up_days",
        "overall_survival_days",
        "overall_survival_event"
    ]
].head(10)
```

```python
print(
    "Patients with OS time:",
    clinical_patient["overall_survival_days"].notna().sum()
)

print(
    "Patients with OS event:",
    clinical_patient["overall_survival_event"].notna().sum()
)

survival_complete = clinical_patient[
    clinical_patient["overall_survival_days"].notna()
    & clinical_patient["overall_survival_event"].notna()
].copy()

print("Patients usable for survival analysis:", len(survival_complete))
```

```python
clinical_patient["overall_survival_days"].describe()
```

```python
print(
    "Negative OS times:",
    (clinical_patient["overall_survival_days"] < 0).sum()
)
```

```python
final_columns = [
    "case_id",
    "patient_id",
    "sex",
    "race",
    "ethnicity",

    "age_at_diagnosis_days",
    "age_at_diagnosis_years",
    "primary_diagnosis",
    "stage",
    "pathologic_T",
    "pathologic_N",
    "pathologic_M",

    "vital_status",
    "days_to_death",
    "latest_follow_up_days",
    "overall_survival_days",
    "overall_survival_event"
]

clinical_clean = clinical_patient[final_columns].copy()

clinical_clean.head()
```

```python
print("Final dataset shape:", clinical_clean.shape)
print("Unique patients:", clinical_clean["patient_id"].nunique())
print("Duplicated patients:", clinical_clean["patient_id"].duplicated().sum())

print(
    "Patients usable for survival:",
    clinical_clean[
        clinical_clean["overall_survival_days"].notna()
        & clinical_clean["overall_survival_event"].notna()
    ].shape[0]
)
```
